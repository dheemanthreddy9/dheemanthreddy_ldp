```python

from sentence_transformers import SentenceTransformer, util
```

```python

sentences = [
    "Plants need water and sunlight to grow.",
    "Flowers require sunlight and water for growth.",
    "Plants are suffering because of a pest attack.",
    "I enjoy watching football matches."
]

for sentence in sentences:
    print(sentence)
```

```python

models = [
    "sentence-transformers/all-MiniLM-L6-v2",
    "sentence-transformers/paraphrase-MiniLM-L3-v2",
    "sentence-transformers/multi-qa-MiniLM-L6-cos-v1"
]

print("Models selected:", len(models))
```

```python

for model_name in models:
    print("\nModel:", model_name)


    model = SentenceTransformer(model_name)


    embeddings = model.encode(sentences)

    print("Embedding shape:", embeddings.shape)


    similarity = util.cos_sim(embeddings, embeddings)

    print("Plants vs Flowers:",
          round(similarity[0][1].item(), 3))

    print("Plants vs Pest Attack:",
          round(similarity[0][2].item(), 3))

    print("Plants vs Football:",
          round(similarity[0][3].item(), 3))
```
